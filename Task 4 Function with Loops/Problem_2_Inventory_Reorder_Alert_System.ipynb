##Problem 2: Inventory Reorder Alert System

```python
def check_inventory(stock_data):
  for product, quantity in stock_data.items():
    if quantity < 15:
      print(f"{product}: Reorder Alert")
    else:
      print(f"{product}: Stock ok")
```

```python
#Example data
inventory = {
    "Apple": 20,
    "Oranges": 10,
    "milk": 25,
    "Bread": 12
}
```

```python
check_inventory(inventory)
```
