##Problem 3: Student Result Processing System

```python
def student_result(marks):
  total = 0

  #calculate total using loop
  for mark in marks:
    total += mark

  average = total / len(marks)

  print("Average marks:", average)

  if average >= 50:
    return "Pass"
  else:
    return "Fail"
```

```python
#Example
marks_list = [60, 70, 45,80]
result = student_result(marks_list)
print("Result:", result)
```
