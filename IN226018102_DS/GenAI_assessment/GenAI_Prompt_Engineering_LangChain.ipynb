```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

```python
# Task 1: Replace Hardcoded Prompt

simple_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners."
)

print(simple_template.format(topic="Artificial Intelligence"))
```

```python
# Task 2: Multi-Input Prompt System

multi_input_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone."
)

print(multi_input_template.format(
    topic="AI",
    audience="beginners",
    tone="friendly"
))

print(multi_input_template.format(
    topic="Python",
    audience="kids",
    tone="fun"
))

print(multi_input_template.format(
    topic="Deep Learning",
    audience="engineers",
    tone="technical"
))
```

```python
# Task 3: Prompt Variations Engine

teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step."
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}."
)

story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story."
)

topic_name = "Machine Learning"

print("Teaching Prompt:")
print(teaching_template.format(topic=topic_name))

print("\nInterview Prompt:")
print(interview_template.format(topic=topic_name))

print("\nStorytelling Prompt:")
print(story_template.format(topic=topic_name))
```

```python
# Task 4: ChatPromptTemplate System

teacher_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful teacher."),
    ("user", "Teach me about {topic}.")
])

interviewer_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an interviewer."),
    ("user", "Ask me questions about {topic}.")
])

motivator_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a motivator."),
    ("user", "Motivate me to learn {topic}.")
])

print("Teacher Role:\n")
print(teacher_prompt.format_messages(topic="Neural Networks"))

print("\nInterviewer Role:\n")
print(interviewer_prompt.format_messages(topic="Neural Networks"))

print("\nMotivator Role:\n")
print(motivator_prompt.format_messages(topic="Neural Networks"))
```

```python
# Task 5: Input Validation Layer

def validate_inputs(audience, tone):

    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        print("Invalid audience. Defaulting to 'beginner'")
        audience = "beginner"

    if tone not in valid_tone:
        print("Invalid tone. Defaulting to 'formal'")
        tone = "formal"

    return audience, tone


# Test validation
audience, tone = validate_inputs("kids", "friendly")
print(audience, tone)
```

```python
# Task 6: Prompt Generator App

def generate_prompt(topic, audience, tone, style):

    audience, tone = validate_inputs(audience, tone)

    templates = {
        "teaching": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} for {audience} in a {tone} teaching style."
        ),

        "interview": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Ask interview questions about {topic} for {audience} in a {tone} style."
        ),

        "storytelling": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} for {audience} in a {tone} storytelling style."
        )
    }

    return templates[style].format(
        topic=topic,
        audience=audience,
        tone=tone
    )

# Example
print(generate_prompt(
    topic="Neural Networks",
    audience="beginner",
    tone="fun",
    style="storytelling"
))
```

```python
# Task 7: Template Reusability Test

reuse_template = PromptTemplate(
    input_variables=["topic", "audience"],
    template="Explain {topic} to {audience} in simple language."
)

test_inputs = [
    {"topic": "AI", "audience": "students"},
    {"topic": "Python", "audience": "kids"},
    {"topic": "Cloud Computing", "audience": "beginners"},
    {"topic": "Data Science", "audience": "professionals"},
    {"topic": "Machine Learning", "audience": "engineers"}
]

for item in test_inputs:
    print(reuse_template.format(
        topic=item["topic"],
        audience=item["audience"]
    ))
    print("-" * 50)
```
