```
import os
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_API_KEY"] = "your_langsmith_api_key"
os.environ["OPENAI_API_KEY"] = "your_openai_api_key"
```

```
job_description = '''
Data Scientist Role
Required Skills:
- Python
- Machine Learning
- SQL
- NLP
- Deep Learning
'''
print(job_description)
```

```
strong_resume = 'Python, SQL, NLP, Deep Learning, 3 years experience'
average_resume = 'Python, SQL, 1 year experience'
weak_resume = 'MS Excel, Fresher'
```

```
from langchain_core.prompts import PromptTemplate

extract_prompt = PromptTemplate(
    input_variables=["resume"],
    template="Extract skills and experience from: {resume}"
)

match_prompt = PromptTemplate(
    input_variables=["resume_data", "job_description"],
    template="Compare {resume_data} with {job_description} and give score."
)
```

```
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model='gpt-3.5-turbo', temperature=0)

extract_chain = extract_prompt | llm
match_chain = match_prompt | llm
```
